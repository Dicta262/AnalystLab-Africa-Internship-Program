# ETL: Data Pipelines & Automation


## Data Extraction

```python
import requests
import csv
from datetime import datetime

# Replace with your actual API key
API_KEY = "a754b7e120ad4a65b591ffc5b154547b"

# Cities you want to check
cities = ["Abuja", "Lagos", "Port Harcourt"]

# Create or overwrite CSV file
with open("nigeria_weather.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    # Write header row
    writer.writerow(["Date", "Time", "City", "Temperature (°C)", "Feels Like (°C)",
                     "Humidity (%)", "Condition", "Wind Speed (m/s)"])

    for city in cities:
        # Build request URL
        url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={API_KEY}&units=metric"
        response = requests.get(url)
        data = response.json()

        # Extract useful info
        temp = data["main"]["temp"]
        feels_like = data["main"]["feels_like"]
        humidity = data["main"]["humidity"]
        description = data["weather"][0]["description"]
        wind_speed = data["wind"]["speed"]

        # Current date and time
        now = datetime.now()
        date_str = now.strftime("%Y-%m-%d")
        time_str = now.strftime("%H:%M:%S")

        # Write row to CSV
        writer.writerow([date_str, time_str, city, temp, feels_like, humidity, description, wind_speed])

print("Weather data saved to nigeria_weather.csv")
```

## Data Transformation

```python
# Import Pandas
import pandas as pd
```

```python
# Load the weather dataset
weather_df = pd.read_csv("nigeria_weather.csv")
```

```python
# Display original data
print("Original Dataset:")
display(weather_df)
```

```python
# Rename columns
weather_df = weather_df.rename(columns={
    "Temperature (°C)": "Temperature_C",
    "Feels Like (°C)": "Feels_Like_C",
    "Humidity (%)": "Humidity_Percent",
    "Wind Speed (m/s)": "Wind_Speed"
})
```

```python
# Convert data types
weather_df["Temperature_C"] = weather_df["Temperature_C"].astype(float)
weather_df["Feels_Like_C"] = weather_df["Feels_Like_C"].astype(float)
weather_df["Humidity_Percent"] = weather_df["Humidity_Percent"].astype(int)
weather_df["Wind_Speed"] = weather_df["Wind_Speed"].astype(float)
```

```python
# Check for missing values
print("\nMissing Values:")
print(weather_df.isnull().sum())
```

```python
# Display cleaned dataset
print("\nCleaned and Transformed Dataset:")
display(weather_df)
```

```python
print("Cleaned and transformed the weather dataset.")
```

```python
# Save cleaned dataset as Excel

weather_df.to_excel("Cleaned_Weather_Dataset.xlsx", index=False)

print("Excel file saved successfully.")
```

## Basic Analysis

```python
# Compare temperatures across cities

print("Temperature Comparison Across Cities")
display(weather_df[["City", "Temperature_C"]])
```

```python
# Find city with highest humidity

highest_humidity_city = weather_df.loc[
    weather_df["Humidity_Percent"].idxmax(),
    ["City", "Humidity_Percent"]
]

print("\nCity with Highest Humidity:")
print(f"{highest_humidity_city['City']} "
      f"({highest_humidity_city['Humidity_Percent']}%)")
```

```python
# Compare weather conditions

print("\nWeather Conditions by City")
display(weather_df[["City", "Condition"]])
```

```python
# Summary of findings

highest_temp_city = weather_df.loc[
    weather_df["Temperature_C"].idxmax(),
    ["City", "Temperature_C"]
]

print("\nSUMMARY OF FINDINGS")
print(f"• {highest_temp_city['City']} recorded the highest temperature "
      f"at {highest_temp_city['Temperature_C']}°C.")

print(f"• {highest_humidity_city['City']} recorded the highest humidity "
      f"at {highest_humidity_city['Humidity_Percent']}%.")

print("• Weather conditions varied across the selected cities.")
```
